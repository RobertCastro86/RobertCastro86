### 1. Comandos de Voz e Leitura de Texto (Text-to-Speech)
Muitos idosos podem ter dificuldades com digitação em teclados pequenos ou problemas de visão (presbiopia).

*   **Como funciona:** Um botão de microfone em destaque permite ditar a tarefa (ex: "Lembrar de tomar o remédio às 20h").
*   **Feedback:** O app confirma por voz o que foi agendado, garantindo que o usuário saiba que a ação foi concluída com sucesso.

### 2. Interface de Alto Contraste e Fontes Dinâmicas
A sensibilidade ao contraste e a acuidade visual diminuem com a idade.

*   **Como funciona:** O app deve oferecer um modo de 'Alta Legibilidade' com botões grandes, ícones autoexplicativos e texto com alto contraste (ex: preto no branco ou amarelo no preto).
*   **Flexibilidade:** Suporte ao redimensionamento de fonte do sistema, garantindo que o texto não quebre ou desapareça ao ser ampliado.

### Protótipo de Interface de Alto Contraste
O código abaixo gera uma interface simulada com elementos focados em acessibilidade visual.

```python
import ipywidgets as widgets
from IPython.display import display

# Estilização de alto contraste (Fundo preto, texto amarelo/branco)
header = widgets.HTML(
    value="<h1 style='color: #FFFF00; background-color: #000000; padding: 20px; text-align: center;'>MEUS LEMBRETES</h1>"
)

# Botões grandes com fontes grandes
btn_style = """
<style>
.custom-btn {
    height: 80px !important;
    width: 100% !important;
    font-size: 24px !important;
    font-weight: bold !important;
    color: #000000 !important;
    background-color: #FFFF00 !important;
    border: 3px solid #FFFFFF !important;
    margin-bottom: 10px;
}
</style>
"""

display(widgets.HTML(btn_style))

# Criando os botões de tarefa
tarefa_1 = widgets.Button(description="💊 TOMAR REMÉDIO - 08:00")
tarefa_1.add_class("custom-btn")

tarefa_2 = widgets.Button(description="🚶 CAMINHADA - 17:00")
tarefa_2.add_class("custom-btn")

btn_adicionar = widgets.Button(description="➕ ADICIONAR NOVO")
btn_adicionar.style.button_color = '#00FF00'
btn_adicionar.add_class("custom-btn")

# Layout do App
container = widgets.VBox([header, tarefa_1, tarefa_2, widgets.HTML("<br>"), btn_adicionar],
                         layout=widgets.Layout(width='400px', border='5px solid #FFFF00', padding='10px', background_color='#000000'))

display(container)
```
