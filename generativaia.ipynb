### Protótipo de Interface de Alto Contraste
O código abaixo gera uma interface simulada com elementos focados em acessibilidade visual.

```python
import ipywidgets as widgets
from IPython.display import display

# Estilização de alto contraste (Fundo preto, texto amarelo/branco)
header = widgets.HTML(
    value="<h1 style='color: #FFFF00; background-color: #000000; padding: 20px; text-align: center;'>MEUS LEMBRETES</h1>"
)

# Botões grandes com fontes grandes
btn_style = """
<style>
.custom-btn {
    height: 80px !important;
    width: 100% !important;
    font-size: 24px !important;
    font-weight: bold !important;
    color: #000000 !important;
    background-color: #FFFF00 !important;
    border: 3px solid #FFFFFF !important;
    margin-bottom: 10px;
}
</style>
"""

display(widgets.HTML(btn_style))

# Criando os botões de tarefa
tarefa_1 = widgets.Button(description="💊 TOMAR REMÉDIO - 08:00")
tarefa_1.add_class("custom-btn")

tarefa_2 = widgets.Button(description="🚶 CAMINHADA - 17:00")
tarefa_2.add_class("custom-btn")

btn_adicionar = widgets.Button(description="➕ ADICIONAR NOVO")
btn_adicionar.style.button_color = '#00FF00'
btn_adicionar.add_class("custom-btn")

# Layout do App
container = widgets.VBox([header, tarefa_1, tarefa_2, widgets.HTML("<br>"), btn_adicionar],
                         layout=widgets.Layout(width='400px', border='5px solid #FFFF00', padding='10px', background_color='#000000'))

display(container)
```

```python
# Passo 1: Instalar as bibliotecas necessárias
!pip install -q -U google-generativeai

# Passo 2: Importar as bibliotecas
import google.generativeai as genai
import json
import os
from google.colab import userdata

# Passo 3: Configurar a API Key com segurança
# Importante: Vá no ícone de chave (Secrets) à esquerda, adicione GOOGLE_API_KEY e ative o acesso.
try:
    api_key = userdata.get('GOOGLE_API_KEY')
    genai.configure(api_key=api_key)
    print("✅ API Key configurada via Secrets!")
except Exception as e:
    print("❌ Erro: Chave 'GOOGLE_API_KEY' não encontrada nos Secrets do Colab.")

# Passo 4: Configurar o modelo
try:
    model = genai.GenerativeModel('gemini-1.5-flash')

    # Passo 5: Criar um prompt estruturado
    prompt = "Forneça uma descrição curta e prática de uma funcionalidade para um aplicativo de planejamento financeiro em português do Brasil:"

    # Passo 6: Gerar três saídas
    responses = []
    print("Gerando sugestões...")
    for i in range(3):
        response = model.generate_content(
            prompt,
            generation_config=genai.types.GenerationConfig(
                max_output_tokens=50,
                temperature=0.7,
            )
        )
        responses.append(response.text.strip())

    # Passo 7: Organizar e salvar em JSON
    saidas = [{"id": i+1, "texto": texto} for i, texto in enumerate(responses)]
    with open('funcionalidades_financeiro.json', 'w') as f:
        json.dump(saidas, f, indent=2, ensure_ascii=False)

    # Passo 8: Exibir resultados
    print("\n--- Sugestões Geradas ---")
    for saida in saidas:
        print(f"ID {saida['id']}: {saida['texto']}")
except Exception as e:
    print(f"\n❌ Falha na execução: {e}")
```

```python
# Passo 1: Importar bibliotecas
import google.generativeai as genai
import json
from google.colab import userdata

# Passo 2: Configurar API
try:
    api_key = userdata.get('GOOGLE_API_KEY')
    genai.configure(api_key=api_key)
    model = genai.GenerativeModel('gemini-1.5-flash')

    # Passo 3: Prompt estruturado para Gestão de Tempo
    prompt = "Forneça uma descrição curta e prática de uma funcionalidade para um aplicativo de gestão de tempo em português do Brasil:"

    # Passo 4: Gerar 3 descrições
    responses = []
    print("Gerando funcionalidades de gestão de tempo...")
    for i in range(3):
        response = model.generate_content(
            prompt,
            generation_config=genai.types.GenerationConfig(max_output_tokens=50, temperature=0.7)
        )
        responses.append(response.text.strip())

    # Passo 5: Salvar em JSON
    saidas_tempo = [{"id": i+1, "texto": texto} for i, texto in enumerate(responses)]
    with open('funcionalidades_tempo.json', 'w') as f:
        json.dump(saidas_tempo, f, indent=2, ensure_ascii=False)

    # Passo 6: Exibir resultados
    print("\n--- Arquivo 'funcionalidades_tempo.json' criado com sucesso! ---")
    for item in saidas_tempo:
        print(f"ID {item['id']}: {item['texto']}")

except Exception as e:
    print(f"Erro: {e}. Certifique-se de configurar a GOOGLE_API_KEY nos Secrets.")
```
